
# Simple Linear Regression – Salary Prediction

## Project Overview
This project uses Simple Linear Regression to predict an employee's salary based on their years of work experience.

## Objective
To understand the relationship between years of experience and salary and build a machine learning model to predict salary.

## Algorithm Used
Simple Linear Regression

## Dataset
Salary_Data_1.csv

## Input Feature
- YearsExperience

## Target Variable
- Salary


## 1. Import Required Libraries

Import the necessary Python libraries for data handling, model building, evaluation and visualization.

```python
import pandas as pd
```

```python
df = pd.read_csv('Salary_Data_1.csv')
```

```python
df.head()
```

## 2. Selecting Input and Target Variables

- X (Input): YearsExperience
- y (Target): Salary

The model uses years of experience to predict an employee's salary.

```python
X = df[['YearsExperience']]      #2D/ Multiple cols
y = df['Salary']          #1D / one col
```

## 3. Building and Training the Model

Create a Linear Regression model and train it using the input features (X) and target variable (y).

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()     #empty model
model.fit(X,y)
```

## 4. Model Coefficients and Intercept

The coefficient shows how much the predicted salary changes when years of experience increases by one year.

The intercept represents the predicted salary when years of experience is zero.

```python
model.coef_ , model.intercept_
```

## 5. Salary Prediction

Use the trained model to predict the salary of an employee with 3 years of work experience.

```python
new_data = pd.DataFrame({'YearsExperience': [3]})

predicted_salary = model.predict(new_data)

print("Predicted Salary:", predicted_salary[0])
```

```python
yp = model.predict(X)
ya = df['Salary']
```

## 6. Model Evaluation

Evaluate the performance of the Linear Regression model using:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- R² Score

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
y_pred = model.predict(X)

mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

## 7. Data Visualization

Visualize the relationship between years of experience and salary using a regression line.

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='blue', label='Actual Salary')

plt.plot(X, model.predict(X), color='red', label='Regression Line')

plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary Prediction using Simple Linear Regression')

plt.show()
```
