```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
X = df[['Age','EstimatedSalary']]
y = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=47)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.svm import SVC
model_poly2 = SVC(kernel='poly',degree=2)
model_poly2.fit(X_train_scaled, y_train)
model_poly2.score(X_test_scaled,y_test)
```

```python
from sklearn.svm import SVC
model_poly = SVC(kernel='poly',degree=3)
model_poly.fit(X_train_scaled, y_train)
model_poly.score(X_test_scaled,y_test)
```

```python
from sklearn.svm import SVC
model_svmlin = SVC(kernel='linear')
model_svmlin.fit(X_train_scaled, y_train)
model_svmlin.score(X_test_scaled,y_test)
```

```python
from sklearn.svm import SVC
model_rbf = SVC(kernel='rbf')
model_rbf.fit(X_train_scaled, y_train)
model_rbf.score(X_test_scaled,y_test)
```

```python
A = np.linspace(0,1,100)
grid=[]
for i in A:
    for j in A:
        grid.append([i,j])
grid = np.array(grid)
grid.shape
```

```python
yp = model_poly2.predict(grid)
plt.scatter(grid[:,0],grid[:,1],c=yp,cmap='RdYlBu',alpha=0.18,s=10)
plt.scatter(X_test_scaled[:,0],X_test_scaled[:,1],c=y_test,cmap='RdYlBu',s=20)
```

```python
yp = model_poly.predict(grid)
plt.scatter(grid[:,0],grid[:,1],c=yp,cmap='RdYlBu',alpha=0.18,s=10)
plt.scatter(X_test_scaled[:,0],X_test_scaled[:,1],c=y_test,cmap='RdYlBu',s=20)
```

```python
yp = model_svmlin.predict(grid)
plt.scatter(grid[:,0],grid[:,1],c=yp,cmap='RdYlBu',alpha=0.18,s=10)
plt.scatter(X_test_scaled[:,0],X_test_scaled[:,1],c=y_test,cmap='RdYlBu',s=20)
```

```python
yp = model_rbf.predict(grid)
plt.scatter(grid[:,0],grid[:,1],c=yp,cmap='RdYlBu',alpha=0.18,s=10)
plt.scatter(X_test_scaled[:,0],X_test_scaled[:,1],c=y_test,cmap='RdYlBu',s=20)
```
